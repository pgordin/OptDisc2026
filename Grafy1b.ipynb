# Funkcje grafowe

```python
def print_matrix(vertices, matrix):
  """
  Wypisuje na ekranie graf, podany jako macierz sąsiedztwa
  """
  n = len(matrix)
  if (vertices is None) or (len(vertices) != n):
    vv = range(1, n+1)
  else:
    vv = vertices
  for i in range(n):
    print(vv[i], ": ", end="", sep="")
    for j in range(n):
      if matrix[i, j]:
        print(" ", vv[j], sep="", end="")
    print("")
```

```python
def print_graph(graph):
  """
  Wypisuje graf w postaci słownika/listy sąsiedztwa.
  """
  for v in graph:
    print(v, ":", sep="", end="")
    for u in graph[v]:
      print(" ", u, sep="", end="")
    print("")
```

## Modyfikacje grafów

```python
def add_vertex(graph, vertex):
  """
  Dodaje wierzchołek do grafu.
  """
  if vertex not in graph:
    graph[vertex] = []

def add_arc(graph, arc):
  """
  Dodaje łuk (skierowany, podany jako para wierzchołków) do grafu (prostego).
  """
  u, v = arc
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)


def add_edge(graph, edge):
  """
  Dodaje krawędź (nieskierowany, podany jako para wierzchołków) do grafu (prostego).
  """
  u, v = edge
  add_vertex(graph, u)
  add_vertex(graph, v)
  if u == v:
    raise ValueError("Graf nieskierowany nie może mieć pętli!")
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

## Generowanie grafów losowych w modeli $G(n,p)$

```python
# dla powtarzalności
seed(2026)
```

```python
# losujemy graf G(10, 1/3)
n = 10
p = 1/3
random_graph = {}
for i in range(1, n+1):
  add_vertex(random_graph, i)
  for j in range(1, i):
    if random() < p:
      add_edge(random_graph, (i, j))
```

```python
print_graph(random_graph)
```
