```python
import numpy as np
import itertools

def algorytm_wegierski(matrix):
    C = matrix.copy().astype(float)
    n = C.shape[0]

    # KROK 1: Redukcja wierszy
    for i in range(n):
        C[i] -= np.min(C[i])

    # KROK 2: Redukcja kolumn
    for j in range(n):
        C[:, j] -= np.min(C[:, j])

    # Główna pętla korekt macierzy
    while True:
        # KROK 3: Szukanie minimalnej liczby linii pokrywających wszystkie zera
        najlepsze_linie = n
        optymalne_wiersze = [False] * n
        optymalne_kolumny = [False] * n
        
        # Generujemy automatycznie wszystkie kombinacje linii dla macierzy o rozmiarze n
        # itertools.product([False, True], repeat=n) tworzy kombinacje prawda/fałsz dla n elementów
        for w_linie in itertools.product([False, True], repeat=n):
            for k_linie in itertools.product([False, True], repeat=n):
                
                # Sprawdzamy, czy ten układ linii pokrywa wszystkie zera
                wszystkie_zera_pokryte = True
                for i in range(n):
                    for j in range(n):
                        if C[i, j] == 0 and not w_linie[i] and not k_linie[j]:
                            wszystkie_zera_pokryte = False
                            break
                    if not wszystkie_zera_pokryte:
                        break
                
                if wszystkie_zera_pokryte:
                    liczba_linii = sum(w_linie) + sum(k_linie)
                    if liczba_linii < najlepsze_linie:
                        najlepsze_linie = float(liczba_linii)
                        optymalne_wiersze = w_linie
                        optymalne_kolumny = k_linie

        # Jeśli potrzebujemy tylu linii, ile wynosi rozmiar macierzy (n), układ zer jest optymalny
        if najlepsze_linie >= n:
            break

        # KROK 4: Modyfikacja macierzy o wartość k
        niezakryte = []
        for i in range(n):
            if not optymalne_wiersze[i]:
                for j in range(n):
                    if not optymalne_kolumny[j]:
                        niezakryte.append(C[i, j])
        
        k = min(niezakryte)

        # Aktualizacja komórek macierzy roboczej
        for i in range(n):
            for j in range(n):
                if not optymalne_wiersze[i] and not optymalne_kolumny[j]:
                    C[i, j] -= k  # Odejmij od niezakrytych
                if optymalne_wiersze[i] and optymalne_kolumny[j]:
                    C[i, j] += k  # Dodaj na skrzyżowaniach linii

    # FINAŁ: Znajdowanie unikalnego przydziału zer (Skojarzenie doskonałe)
    for p in itertools.permutations(range(n)):
        if all(C[i, p[i]] == 0 for i in range(n)):
            return np.arange(n), np.array(p)


# URUCHOMIENIE

if __name__ == "__main__":
    # Przykładowa macierz 
    macierz_kosztow = np.array([
        [10, 15, 9],
        [9, 18, 5],
        [6, 14, 3]
    ])

    r_ind, c_ind = algorytm_wegierski(macierz_kosztow)

    print("Poprawny przydział dla macierzy o dowolnym rozmiarze:")
    for r, c in zip(r_ind, c_ind):
        print(f"  Pracownik {r} ---> Zadanie {c} | Koszt: {macierz_kosztow[r, c]} PLN")
        
    total_cost = macierz_kosztow[r_ind, c_ind].sum()
    print("-" * 50)
    print(f"Minimalny koszt całkowity: {total_cost} PLN")
```
