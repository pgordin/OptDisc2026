```python
#zadanie 1
from collections import deque

class Graph:

    def __init__(self):
        self.graph = {}

    def addEdge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)
        self.graph[v].append(u)

    def ConnectedComponentsBFS(self):

        visited = set()
        components = []

        for start in self.graph:

            if start not in visited:

                component = set()
                queue = deque([start])

                visited.add(start)

                while queue:

                    v = queue.popleft()
                    component.add(v)

                    for u in self.graph[v]:

                        if u not in visited:
                            visited.add(u)
                            queue.append(u)

                components.append(component)

        return components


G = Graph()

G.addEdge(1, 2)
G.addEdge(2, 3)
G.addEdge(4, 5)

print(G.ConnectedComponentsBFS())
```

```python
#zadanie 2
class Graph:

    def __init__(self):
        self.graph = {}

    def addEdge(self, u, v):

        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)

    def TopologicalSort(self):

        visited = set()
        result = []

        def DFS(v):

            visited.add(v)

            for u in self.graph[v]:
                if u not in visited:
                    DFS(u)

            result.append(v)

        for v in self.graph:
            if v not in visited:
                DFS(v)

        result.reverse()

        return result


G = Graph()

G.addEdge(5, 2)
G.addEdge(5, 0)
G.addEdge(4, 0)
G.addEdge(4, 1)
G.addEdge(2, 3)
G.addEdge(3, 1)

print(G.TopologicalSort())
```

```python
#zadanie 3
class Graph:

    def __init__(self):
        self.graph = {}

    def addEdge(self, u, v):

        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)

    def StronglyConnectedComponents(self):

        visited = set()
        order = []

        # ---------- pierwszy DFS ----------
        def DFS1(v):

            visited.add(v)

            for u in self.graph[v]:
                if u not in visited:
                    DFS1(u)

            order.append(v)

        for v in self.graph:
            if v not in visited:
                DFS1(v)

        # ---------- odwrócenie grafu ----------
        reversed_graph = {}

        for v in self.graph:
            reversed_graph[v] = []

        for v in self.graph:
            for u in self.graph[v]:
                reversed_graph[u].append(v)

        # ---------- drugi DFS ----------
        visited.clear()
        components = []

        def DFS2(v, component):

            visited.add(v)
            component.append(v)

            for u in reversed_graph[v]:
                if u not in visited:
                    DFS2(u, component)

        while order:

            v = order.pop()

            if v not in visited:

                component = []
                DFS2(v, component)

                components.append(component)

        return components


G = Graph()

G.addEdge(0, 1)
G.addEdge(1, 2)
G.addEdge(2, 0)

G.addEdge(2, 3)

G.addEdge(3, 4)

print(G.StronglyConnectedComponents())
```
