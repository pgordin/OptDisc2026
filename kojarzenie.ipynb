```python
from collections import deque
class HopcroftKarp:
    def __init__(self, boys, girls):
        self.boys = boys
        self.girls = girls

        self.graph = {boy: [] for boy in boys}

        self.pair_boy = {}
        self.pair_girl = {}

        self.dist = {}

    def add_edge(self, boy, girl):
        self.graph[boy].append(girl)

    def bfs(self):
        queue = deque()
        for boy in self.boys:
            if boy not in self.pair_boy:
                self.dist[boy] = 0
                queue.append(boy)
            else:
                self.dist[boy] = float('inf')
        found = False
        while queue:
            boy = queue.popleft()
            for girl in self.graph[boy]:
                if girl not in self.pair_girl:
                    found = True
                else:
                    next_boy = self.pair_girl[girl]
                    if self.dist[next_boy] == float('inf'):
                        self.dist[next_boy] = self.dist[boy] + 1
                        queue.append(next_boy)

        return found

    def dfs(self, boy):
        for girl in self.graph[boy]:
            if girl not in self.pair_girl:
                self.pair_boy[boy] = girl
                self.pair_girl[girl] = boy
                return True
            next_boy = self.pair_girl[girl]
            if self.dist[next_boy] == self.dist[boy] + 1:
                if self.dfs(next_boy):
                    self.pair_boy[boy] = girl
                    self.pair_girl[girl] = boy
                    return True

        self.dist[boy] = float('inf')
        return False

    def maximum_matching(self):
        matching = 0
        while self.bfs():
            for boy in self.boys:
                if boy not in self.pair_boy:
                    if self.dfs(boy):
                        matching += 1
        return matching

    def print_matching(self):
        print("\nNajwiększe skojarzenie:\n")
        for boy in self.pair_boy:
            print(f"{boy}  ->  {self.pair_boy[boy]}")

        print(f"\nLiczba par: {len(self.pair_boy)}")



boys = ["Adam", "Bartek", "Kuba", "Michał", "Krzyś"]
girls = ["Ola", "Zosia", "Basia", "Natalka"]

hk = HopcroftKarp(boys, girls)

hk.add_edge("Adam", "Ola")
hk.add_edge("Adam", "Zosia")

hk.add_edge("Bartek", "Ola")

hk.add_edge("Kuba", "Basia")
hk.add_edge("Kuba", "Zosia")

hk.add_edge("Michał", "Basia")
hk.add_edge("Michał", "Natalka")

hk.add_edge("Krzyś", "Ola")
hk.add_edge("Krzyś", "Basia")

max_pairs = hk.maximum_matching()
hk.print_matching()
```
