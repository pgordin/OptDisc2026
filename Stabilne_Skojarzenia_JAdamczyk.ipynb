```python
def gale_shapley(preferencje_A, preferencje_B):
    """
    Uniwersalna implementacja algorytmu Gale'a-Shapleya.
    Obsługuje: pełne grafy, nierówne zbiory oraz grafy niepełne (brakujące krawędzie).
    """
    wolni_A = list(preferencje_A.keys())
    nastepna_propozycja = {a: 0 for a in preferencje_A}
    parowanie_B = {b: None for b in preferencje_B}
    
    # Tworzenie odwróconej macierzy dla szybkiego sprawdzania preferencji w czasie O(1)
    rank_B = {}
    for b, lista_preferencji in preferencje_B.items():
        rank_B[b] = {a: indeks for indeks, a in enumerate(lista_preferencji)}
        
    while wolni_A:
        a = wolni_A[0]
        indeks_kandydata = nastepna_propozycja[a]
        
        # Zabezpieczenie na wypadek nierównych zbiorów lub niepełnych list
        if indeks_kandydata >= len(preferencje_A[a]):
            wolni_A.pop(0) # Element 'a' wyczerpał listę i trwale zostaje bez pary
            continue
            
        b = preferencje_A[a][indeks_kandydata]
        nastepna_propozycja[a] += 1
        
        stary_partner = parowanie_B.get(b)
        
        if stary_partner is None:
            parowanie_B[b] = a
            wolni_A.pop(0)
        else:
            pozycja_nowego = rank_B[b].get(a, float('inf'))
            pozycja_starego = rank_B[b].get(stary_partner, float('inf'))
            
            if pozycja_nowego < pozycja_starego:
                parowanie_B[b] = a
                wolni_A.pop(0)
                wolni_A.append(stary_partner)
                
    return {a: b for b, a in parowanie_B.items() if a is not None}


# ==========================================
# PREZENTACJA WYNIKÓW - 3 WARIANTY GRAFÓW
# ==========================================

print("="*60)
print("PRZYPADEK 1: Równe zbiory, Pełny graf (3 na 3)")
print("Oczekiwany wynik: Wszyscy znajdują parę.")
print("="*60)
pref_A1 = {
    'A1': ['B1', 'B2', 'B3'], 
    'A2': ['B1', 'B3', 'B2'], 
    'A3': ['B2', 'B1', 'B3']
}
pref_B1 = {
    'B1': ['A2', 'A3', 'A1'], 
    'B2': ['A1', 'A2', 'A3'], 
    'B3': ['A1', 'A2', 'A3']
}
wynik1 = gale_shapley(pref_A1, pref_B1)
for a in sorted(wynik1.keys()): 
    print(f"{a} <-> {wynik1[a]}")


print("\n" + "="*60)
print("PRZYPADEK 2: Nierówne zbiory (3 kandydatów, 2 miejsca)")
print("Oczekiwany wynik: Brak miejsc, jeden kandydat (A3) zostaje singlem.")
print("="*60)
pref_A2 = {
    'A1': ['B1', 'B2'], 
    'A2': ['B1', 'B2'], 
    'A3': ['B2', 'B1']
}
pref_B2 = {
    'B1': ['A1', 'A2', 'A3'], 
    'B2': ['A2', 'A1', 'A3']
}
wynik2 = gale_shapley(pref_A2, pref_B2)
for a in sorted(wynik2.keys()): 
    print(f"{a} <-> {wynik2[a]}")
    
nie_znalalezli_pracy = set(pref_A2.keys()) - set(wynik2.keys())
print(f"Pozostali singlem: {', '.join(nie_znalalezli_pracy)}")


print("\n" + "="*60)
print("PRZYPADEK 3: Graf niepełny (3 na 3, brakujące krawędzie)")
print("Sytuacja: A1 zgadza się tylko na B1. B1 woli A2. A1 odpada.")
print("Oczekiwany wynik: Równe zbiory, a mimo to A1 i B3 zostają sami.")
print("="*60)
pref_A3 = {
    'A1': ['B1'],               # A1 chce TYLKO B1, inne stanowiska odrzuca
    'A2': ['B1', 'B2'], 
    'A3': ['B2', 'B3']
}
pref_B3 = {
    'B1': ['A2', 'A1'], 
    'B2': ['A3', 'A2'], 
    'B3': ['A3']
}
wynik3 = gale_shapley(pref_A3, pref_B3)
for a in sorted(wynik3.keys()): 
    print(f"{a} <-> {wynik3[a]}")
    
single_A = set(pref_A3.keys()) - set(wynik3.keys())
single_B = set(pref_B3.keys()) - set(wynik3.values())
print(f"Zostali bez przypisania:")
print(f" - Kandydaci: {', '.join(single_A)}")
print(f" - Stanowiska: {', '.join(single_B)}")
```
