```python
# @title
#zad2
def graph_to_matrix(graph):
    vertices = list(graph.keys()) #wierzchołki grafu
    n = len(vertices) #liczba wierszy

    index = {v: i for i, v in enumerate(vertices)}
    matrix = [[0] * n for _ in range(n)] #pusta macierz

    for v in graph:
        for u in graph[v]:
            i, j = index[v], index[u]
            matrix[i][j] = 1

    return vertices, matrix
#zad3
def matrix_to_graph(vertices, matrix):
    graph = {v: [] for v in vertices}
    n = len(vertices)
    if len(matrix) != n or any(len(row) != n for row in matrix):
        raise ValueError("liczba wierzchołków nie pasuje do wielkości macierzy")
    for i in range(n):
        for j in range(n):
            if matrix[i][j] == 1:
                graph[vertices[i]].append(vertices[j])

    return graph
def add_vertex(graph, vertex):
  """
  Dodaje wierzchołek do grafu
  """
  if vertex not in graph:
    graph[vertex] = []

def add_arc(graph, arc):
  """
  Dodaje łuk (skierowany, podany jako para wierzchołków) do grafu
  """
  u, v = arc
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)

def add_edge(graph, edge):
  """
  Dodaje krawędź (podaną jako para wierzchołków) do grafu
  Rozpatrujemy grafy proste, nieskierowane
  """
  u, v = edge
  add_vertex(graph, u)
  add_vertex(graph, v)
  if u == v:
    raise ValueError("Pętla!!!")
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

Dwie funkcje do zapisu grafów z zajęć.

```python
def graph_from_edges(filename, directed=0):
  """
  Tworzy graf na podstawie pliku z krawędziami. Opis krawędzi to dwa słowa lub wierzchołka (jedno słowo).
  Nadmiarowe słowa są ignorowane.
  Zmienna filename zawiera pełną ścieżkę pliku
  """
  graph = {}
  with open(filename, 'r') as file: # otwarcie do odczytu
    for line in file:
      words = line.strip().split()
      if len(words) == 1:     # jedno słowo - opis wierzchołka
        add_vertex(graph, words[0])
      elif len(words) >= 2:   # co najmniej dwa słowa - opis krawędzi
        if directed:
          add_arc(graph, (words[0], words[1]))
        else:
          add_edge(graph, (words[0], words[1]))
  return graph
```

```python
def graph_to_neighbourlist(graph, filename):
  """
  Zapisuje graf podany jako słownik (list) sąsiedztwa do pliku (w formie listy sąsiedztwa).
  Zmienna filename powinna zawierać pełną ścieżkę do pliku.
  """


  with open(filename, 'w') as file: # otwarcie do zapisu
    for v in graph:
      line = f"{v}:"
      for u in graph[v]:
        line += f" {u}"
      line += "\n"
      file.write(line)
```

funkcje brakujące

```python
def graph_to_edges(graph, filename, directed):
  """
  Zapisuje graf do pliku w formacie listy krawędzi.
  Każda krawędź w osobnym wierszu: u v
  u i v to wierzchołki

  """
  with open(filename, 'w') as file:
    visited = set()  # ma nie dublować krawędzi w grafach nieskierowanych

    for u in graph: #dwa przypadki dla skierowanych i nieskierowanych
      for v in graph[u]:
        if directed:
          file.write(f"{u} {v}\n")
        else:
          if (v, u) not in visited:
            file.write(f"{u} {v}\n")
            visited.add((u, v))
```

```python
def graph_from_neighbourlist(filename, directed=0):
  """
  Tworzy graf na podstawie pliku z listą sąsiedztwa.
  Format: v: u1 u2 u3
  filename musi zawierać dobrą ściezkę do pliku.
  """
  graph = {}

  try:
    with open(filename, 'r') as file:
      for line in file:
        parts = line.strip().split(':')

        if len(parts) != 2:
          raise ValueError(f"Błędny format linii: {line.strip()}")

        v = parts[0].strip()
        neighbours = parts[1].strip().split()

        add_vertex(graph, v)

        for u in neighbours:
          if directed:
            add_arc(graph, (v, u))
          else:
            add_edge(graph, (v, u))

  except FileNotFoundError:
    raise FileNotFoundError(f"Plik '{filename}' nie istnieje.")

  except Exception as e: #jakiś inny błąd
    raise Exception(f"Błąd podczas wczytywania grafu: {e}")

  return graph
```
