# Zadanie 1.
Na przykładzie ciągłego problemu plecakowego sprawdzić (i przetestować) SciPy, PuLP i Google OR-Tools.

- - -

```python
# Google OR-Tools
from ortools.linear_solver import pywraplp



names = ["A","B","C","D","E","F"]
values = [5000,1500,200,1000,500,600]
weights = [1,0.5,0.3,4,1,3]
max_items = [1,3,5,1,5,2]

capacity = 11.5

# solver
solver = pywraplp.Solver.CreateSolver('GLOP')

# zmienne
x = []
for i in range(len(names)):
    x.append(solver.NumVar(0, max_items[i], names[i]))

# ograniczenie wagowe
solver.Add(sum(weights[i]*x[i] for i in range(len(names))) <= capacity)

# funkcja celu
solver.Maximize(sum(values[i]*x[i] for i in range(len(names))))

status = solver.Solve()

if status == pywraplp.Solver.OPTIMAL:
    print("Optymalna wartość =", solver.Objective().Value())
    print()

    total_weight = 0

    for i in range(len(names)):
        val = x[i].solution_value()
        print(names[i], "=", val)
        total_weight += val * weights[i]

    print()
    print("Łączna waga =", total_weight)
else:
  print("Problem ten nie ma optymalnego rozwiązania.")
```

# Zadanie 2.
Jak wygląda programowanie całkowitoliczbowe w tych pakietach? Rozwiązać (zwykły) problem plecakowy.


```python
from ortools.linear_solver import pywraplp

names = ["A","B","C","D","E","F"]
values = [5000,1500,200,1000,500,600]
weights = [1,0.5,0.3,4,1,3]
max_items = [1,3,5,1,5,2]

capacity = 11.5

# solver dla programowania całkowitoliczbowego
solver = pywraplp.Solver.CreateSolver('SCIP')

# zmienne całkowite
x = []
for i in range(len(names)):
    x.append(solver.IntVar(0, max_items[i], names[i]))

# ograniczenie wagowe
solver.Add(sum(weights[i] * x[i] for i in range(len(names))) <= capacity)

# funkcja celu
solver.Maximize(sum(values[i] * x[i] for i in range(len(names))))

status = solver.Solve()

if status == pywraplp.Solver.OPTIMAL:

    print("Optymalna wartość =", solver.Objective().Value())
    print()

    total_weight = 0

    for i in range(len(names)):
        val = x[i].solution_value()
        print(names[i], "=", val)
        total_weight += val * weights[i]

    print()
    print("Łączna waga =", total_weight)

else:
    print("Problem nie ma optymalnego rozwiązania.")
```
