```python
import numpy as np
from random import random, seed
```

```python
#zad1 losowy graf nieskierowany
seed(2026)

def random_graph(n, p):
    graph = {i: [] for i in range(n)}

    for i in range(n):
        for j in range(i + 1, n):
            if random() < p:
                graph[i].append(j)
                graph[j].append(i)

    return graph

print(random_graph(5,0.3))
```

```python
#zad2
def graph_to_matrix(graph):
    vertices = list(graph.keys()) #wierzchołki grafu
    n = len(vertices) #liczba wierszy

    index = {v: i for i, v in enumerate(vertices)}
    matrix = [[0] * n for _ in range(n)] #pusta macierz

    for v in graph:
        for u in graph[v]:
            i, j = index[v], index[u]
            matrix[i][j] = 1

    return vertices, matrix
```

```python
#zad3
def matrix_to_graph(vertices, matrix):
    graph = {v: [] for v in vertices}
    n = len(vertices)
    if len(matrix) != n or any(len(row) != n for row in matrix):
        raise ValueError("liczba wierzchołków nie pasuje do wielkości macierzy")
    for i in range(n):
        for j in range(n):
            if matrix[i][j] == 1:
                graph[vertices[i]].append(vertices[j])

    return graph
```

```python
#zad4 cykl
def cycle(n):
    graph = {i: [] for i in range(n)}

    for i in range(n):
        graph[i].append((i + 1) % n)
        graph[i].append((i - 1) % n)

    return graph
print(cycle(4))
```
