```
from collections import deque, defaultdict
import matplotlib.pyplot as plt #wykresy
import matplotlib.patches as mpatches #kształty na wykresach
import networkx as nx #grafy
import sys


# STRUKTURY DANYCH

class Graf:
    """
    Skierowany graf ważony przechowywany jako lista sąsiedztwa.
    Wierzchołki numerujemy od 1 do n (jak w treści zadania).
    """

    def __init__(self, n):
        if n < 2:  # graf musi mieć co najmniej dwa wierzchołki
            raise ValueError("Graf musi mieć co najmniej 2 wierzchołki.")
        self.n = n                                          # liczba wierzchołków
        self.krawedzie = defaultdict(list)                  # lista sąsiedztwa: v -> [(u, waga), ...]
        self.wszystkie_krawedzie = []                       # pełna lista krawędzi do rysowania

    def dodaj_krawedz(self, u, v, waga):
        """Dodaje skierowaną krawędź u -> v z podaną wagą."""
        if not (1 <= u <= self.n and 1 <= v <= self.n):    # sprawdzamy poprawność numerów
            raise ValueError(f"Wierzchołek poza zakresem 1..{self.n}.")
        if waga <= 0:                                       # wagi muszą być dodatnie
            raise ValueError("Waga krawędzi musi być dodatnia.")
        self.krawedzie[u].append((v, waga))                 # dopisujemy sąsiada v z wagą
        self.wszystkie_krawedzie.append((u, v, waga))       # dzieki defaultdict jak klucz u nie istnije to dodany automatycznie zapisujemy dla rysunków

    def __str__(self):
        """Tekstowy opis grafu."""
        linie = []
        for v in range(1, self.n + 1):                     # wypisujemy każdy wierzchołek
            sasiedzi = ", ".join(
                f"{u}(waga={w})" for u, w in self.krawedzie[v]
            )
            linie.append(f"  {v}: {sasiedzi if sasiedzi else '—'}")
        return "\n".join(linie)


# KROKI ALGORYTMU

def bfs_osiagalne(graf, start):
    """
    BFS od wierzchołka start. Zwraca zbiór wierzchołków osiągalnych
    ze start (włącznie z nim samym), nie wracając do start jako celu.

    Uwaga: wierzchołek start może być odwiedzony jako punkt pośredni,
    ale nie jako cel ponownego wejścia — symulujemy to, nie traktując
    krawędzi powrotnych DO start jako użytecznych.
    """
    odwiedzone = {start}                                    # zbiór odwiedzonych wierzchołków
    kolejka = deque([start])                                # kolejka BFS

    while kolejka:                                          # dopóki kolejka niepusta
        u = kolejka.popleft()                               # bierzemy pierwszy wierzchołek
        for v, _ in graf.krawedzie[u]:                      # przechodzimy po sąsiadach
            if v != start and v not in odwiedzone:          # pomijamy powrót do domu
                odwiedzone.add(v)                           # zaznaczamy jako odwiedzony
                kolejka.append(v)                           # dodajemy do kolejki

    return odwiedzone                                       # zwracamy zbiór osiągalnych


def wykryj_cykl_dfs(graf, osiagalne):
    """
    Wykrywa cykl w podgrafie indukowanym przez wierzchołki z 'osiagalne'.
    Używa klasycznego DFS z kolorowaniem: biały/szary/czarny.

      - biały (0) = nieodwiedzony,
      - szary  (1) = w aktualnej ścieżce DFS (możliwy cykl),
      - czarny (2) = w pełni przetworzony.

    Uwaga: przed wywołaniem należy usunąć z 'osiagalne' wierzchołki,
    do których powrót jest zakazany (np. dom Jarka).

    Zwraca True, jeśli w podgrafie istnieje cykl.
    """
    kolor = {v: 0 for v in osiagalne}                      # na początku wszystkie białe

    def dfs(u):
        """Rekurencyjny DFS. Zwraca True, gdy znajdzie cykl."""
        kolor[u] = 1                                        # zaznaczamy u jako szary (w trakcie)
        for v, _ in graf.krawedzie[u]:                      # przechodzimy po następnikach u
            if v not in osiagalne:                          # pomijamy wierzchołki spoza podgrafu
                continue
            if kolor[v] == 1:                               # trafiliśmy na szary — cykl!
                return True
            if kolor[v] == 0:                               # biały — eksplorujemy dalej
                if dfs(v):
                    return True
        kolor[u] = 2                                        # u w pełni przetworzony — czarny
        return False                                        # brak cyklu w tej gałęzi

    for v in osiagalne:                                     # startujemy DFS z każdego nieodwiedzonego
        if kolor[v] == 0:
            if dfs(v):
                return True                                 # znaleziono cykl

    return False                                            # brak cyklu w całym podgrafie


def sortowanie_topologiczne(graf, wierzcholki, zablokowany_cel=None, zablokowane_wyjscie=None):
    """
    Sortowanie topologiczne wierzchołków z podanego zbioru.
    Używa algorytmu Kahna (BFS po stopniach wejściowych).

    Parametry:
      zablokowany_cel     — krawędzie WCHODZĄCE do tego wierzchołka są ignorowane
                            (zakaz powrotu do domu).
      zablokowane_wyjscie — krawędzie WYCHODZĄCE z tego wierzchołka są ignorowane
                            (szkoła = węzeł-zlew, Jarek musi się zatrzymać).

    Zwraca listę wierzchołków w porządku topologicznym.
    """
    stopien_wej = defaultdict(int)                          # stopień wejściowy każdego wierzchołka

    for u in wierzcholki:                                   # liczymy stopnie wejściowe
        if u == zablokowane_wyjscie:                        # ze szkoły nie wychodzą krawędzie
            continue
        for v, _ in graf.krawedzie[u]:
            if v in wierzcholki and v != zablokowany_cel:   # ignorujemy krawędzie do zablokowanego
                stopien_wej[v] += 1

    kolejka = deque()                                       # kolejka wierzchołków o stopniu 0
    for v in wierzcholki:
        if stopien_wej[v] == 0:                             # wierzchołki bez poprzedników
            kolejka.append(v)

    porzadek = []                                           # tu zbieramy porządek topologiczny
    while kolejka:
        u = kolejka.popleft()                               # bierzemy wierzchołek o stopniu 0
        porzadek.append(u)                                  # dopisujemy do porządku
        if u == zablokowane_wyjscie:                        # ze szkoły nie wychodzimy dalej
            continue
        for v, _ in graf.krawedzie[u]:                      # przechodzimy po następnikach
            if v in wierzcholki and v != zablokowany_cel:   # ignorujemy krawędzie do zablokowanego
                stopien_wej[v] -= 1                         # zmniejszamy stopień wejściowy
                if stopien_wej[v] == 0:                     # jeśli osiągnął 0 — do kolejki
                    kolejka.append(v)

    return porzadek                                         # lista w porządku topologicznym


def dp_najdluzsza_sciezka(graf, start, koniec, wierzcholki):
    """
    Programowanie dynamiczne na DAG-u w porządku topologicznym.
    Oblicza najdłuższą ścieżkę z 'start' do każdego wierzchołka.

    Krawędzie prowadzące z powrotem DO 'start' są ignorowane —
    Jarek nie może wrócić do domu.

    Krawędzie wychodzące Z 'koniec' są ignorowane —
    gdy Jarek dojeżdża do szkoły, zatrzymuje się.

    dist[v]  = długość najdłuższej ścieżki z start do v
    poprz[v] = poprzednik v na tej ścieżce (do odtworzenia trasy)

    Zwraca (dist, poprz).
    """
    porzadek = sortowanie_topologiczne(                     # topologiczny porządek wierzchołków
        graf, wierzcholki,
        zablokowany_cel=start,                              # zakaz powrotu do domu
        zablokowane_wyjscie=koniec,                         # szkoła = węzeł-zlew
    )

    dist = {v: float('-inf') for v in wierzcholki}          # inicjalizacja: -inf dla wszystkich
    poprz = {v: None for v in wierzcholki}                  # inicjalizacja: brak poprzednika
    dist[start] = 0                                         # odległość do samego startu = 0

    for u in porzadek:                                      # przetwarzamy w porządku topologicznym
        if dist[u] == float('-inf'):                        # jeśli u jest nieosiągalny — pomijamy
            continue
        if u == koniec:                                     # ze szkoły nie idziemy dalej
            continue
        for v, waga in graf.krawedzie[u]:                   # rozważamy każdą krawędź u -> v
            if v not in wierzcholki:                        # pomijamy wierzchołki spoza podgrafu
                continue
            if v == start:                                  # Jarek nie może wrócić do domu!
                continue
            nowa = dist[u] + waga                           # długość ścieżki przez u do v
            if nowa > dist[v]:                              # jeśli dłuższa niż dotychczasowa
                dist[v] = nowa                              # aktualizujemy dist[v]
                poprz[v] = u                                # zapamiętujemy poprzednika

    return dist, poprz                                      # zwracamy słowniki


def odtworz_sciezke(poprz, start, koniec):
    """
    Odtwarza ścieżkę z 'start' do 'koniec' korzystając ze słownika poprzedników.
    Zwraca listę wierzchołków od start do koniec, lub None jeśli brak ścieżki.
    """
    if poprz[koniec] is None and koniec != start:           # brak ścieżki
        return None
    sciezka = []
    v = koniec
    while v is not None:                                    # cofamy się po poprzednikach
        sciezka.append(v)
        v = poprz[v]
    sciezka.reverse()                                       # odwracamy — teraz od start do koniec
    if sciezka[0] != start:                                 # sprawdzamy, czy zaczyna od startu
        return None
    return sciezka                                          # zwracamy ścieżkę


# GŁÓWNA FUNKCJA ROZWIĄZUJĄCA

def rozwiaz(graf, verbose=True):
    """
    Główna funkcja rozwiązująca problem 'Daleka droga do szkoły'.

    Parametry:
      graf    — obiekt klasy Graf
      verbose — czy wypisywać szczegółowy przebieg algorytmu

    Zwraca słownik z wynikiem i szczegółami:
      'wynik'   — 'BRAK', 'WAGARY', lub int (długość najdłuższej drogi)
      'sciezka' — lista wierzchołków najdłuższej drogi (lub None)
      'dist'    — słownik odległości (lub None)
    """
    n = graf.n
    start = 1                                               # dom Jarka
    koniec = n                                              # szkoła

    if verbose:
        print("─" * 60)
        print(f"  Start (dom): {start},  Cel (szkoła): {koniec}")
        print("─" * 60)

    # ── KROK 1: osiągalność ─────────────────────────────────
    if verbose:
        print("\nKROK 1: Sprawdzamy osiągalność szkoly z domu (BFS)")

    osiagalne_z_1 = bfs_osiagalne(graf, start)              # wierzchołki osiągalne z domu

    if verbose:
        print(f"  Osiągalne z {start}: {sorted(osiagalne_z_1)}")

    if koniec not in osiagalne_z_1:                         # szkoła niedostępna
        if verbose:
            print(f"  Wierzchołek {koniec} NIE jest osiągalny z {start}.")
        return {'wynik': 'BRAK', 'sciezka': None, 'dist': None,
                'osiagalne': osiagalne_z_1, 'porzadek_topo': None}

    if verbose:
        print(f"  Wierzchołek {koniec} jest osiągalny — kontynuujemy.")

    # ── KROK 2: wykrycie cyklu ──────────────────────────────
    if verbose:
        print("\nKROK 2: Sprawdzamy, czy w osiągalnym podgrafie jest cykl (DFS kolorowanie)")

    # Jarek nie może ponownie przejechać przez dom (wierzchołek 1) — dom wyłączamy z podgrafu.
    # Gdy Jarek dotrze do szkoły (wierzchołek n), podróż się kończy — szkolę też wyłączamy z podgrafu, bo Jarek musi tam zatrzymać jazdę.
    # Cykl warunkujący WAGARY musi więc istnieć w grafie BEZ wierzchołków 1 i n.
    podgraf_bez_startu = osiagalne_z_1 - {start, koniec}    # podgraf bez domu i szkoły

    jest_cykl = wykryj_cykl_dfs(graf, podgraf_bez_startu)   # sprawdzamy cykl w tym podgrafie

    if verbose:
        if jest_cykl:
            print("  Wykryto cykl w osiągalnym podgrafie → WAGARY")
        else:
            print("  Brak cyklu → podgraf jest DAG-iem, możemy użyć DP.")

    if jest_cykl:
        return {'wynik': 'WAGARY', 'sciezka': None, 'dist': None,
                'osiagalne': osiagalne_z_1, 'porzadek_topo': None}

    # ── KROK 3: DP na DAG-u ─────────────────────────────────
    if verbose:
        print("\nKROK 3: Programowanie dynamiczne na DAG-u")

    wierzcholki_dp = osiagalne_z_1                          # DP na wszystkich osiągalnych (z 1)
    dist, poprz = dp_najdluzsza_sciezka(                    # obliczamy najdłuższe ścieżki
        graf, start, koniec, wierzcholki_dp
    )

    porzadek_topo = sortowanie_topologiczne(                # porządek topologiczny do wyświetlenia
        graf, wierzcholki_dp,
        zablokowany_cel=start,
        zablokowane_wyjscie=koniec,
    )

    if verbose:
        print("  Porządek topologiczny (Kahna):", " → ".join(map(str, porzadek_topo)))
        print("\n  Długości najdłuższych ścieżek z wierzchołka 1:")
        for v in sorted(wierzcholki_dp):
            d = dist[v]
            s = str(d) if d != float('-inf') else "nieosiągalny"
            print(f"    dist[{v}] = {s}")

    # ── KROK 4: odczytanie wyniku ───────────────────────────
    dlugosc = dist[koniec]

    if dlugosc == float('-inf'):                            # niedostępny — nie powinno się zdarzyć
        return {'wynik': 'BRAK', 'sciezka': None, 'dist': dist,
                'osiagalne': osiagalne_z_1, 'porzadek_topo': porzadek_topo}

    sciezka = odtworz_sciezke(poprz, start, koniec)         # odtwarzamy trasę

    if verbose:
        print(f"\n  Najdłuższa droga: {int(dlugosc)} m")
        if sciezka:
            print(f"  Trasa: {' → '.join(map(str, sciezka))}")

    return {
        'wynik': int(dlugosc),
        'sciezka': sciezka,
        'dist': dist,
        'osiagalne': osiagalne_z_1,
        'porzadek_topo': porzadek_topo,
    }


# WCZYTYWANIE Z DANYCH (format zadania)

def wczytaj_z_tekstu(tekst):
    """
    Wczytuje graf z tekstu w formacie zadania:
      pierwsza linia: n m
      kolejne m linii: u v waga
    Zwraca obiekt Graf.
    """
    linie = tekst.strip().splitlines()                      # dzielimy na linie
    n, m = map(int, linie[0].split())                       # pierwsza linia: n i m
    g = Graf(n)                                             # tworzymy pusty graf
    for i in range(1, m + 1):                               # wczytujemy m krawędzi
        u, v, w = map(int, linie[i].split())
        g.dodaj_krawedz(u, v, w)
    return g


# RYSOWANIE GRAFÓW

def _layout(graf):
    """Tworzy rozmieszczenie wierzchołków grafu do rysowania."""
    G = nx.DiGraph()
    for v in range(1, graf.n + 1):
        G.add_node(v)
    for u, v, w in graf.wszystkie_krawedzie:
        G.add_edge(u, v)
    return nx.spring_layout(G, seed=42, k=2.0)             # spring layout z ustalonym seedem


def rysuj_graf_wejsciowy(graf, tytul, polozenie=None):
    """Rysuje wejściowy graf skierowany z wagami na krawędziach."""
    G = nx.DiGraph()
    for v in range(1, graf.n + 1):
        G.add_node(v)
    for u, v, w in graf.wszystkie_krawedzie:
        G.add_edge(u, v, waga=w)

    if polozenie is None:
        polozenie = nx.spring_layout(G, seed=42, k=2.0)

    kolory = []                                             # kolory wierzchołków
    for v in G.nodes():
        if v == 1:
            kolory.append("lightgreen")                     # zielony = dom Jarka
        elif v == graf.n:
            kolory.append("lightcoral")                     # czerwony = szkoła
        else:
            kolory.append("lightyellow")                    # żółty = pośrednie

    etykiety_krawedzi = {
        (u, v): dane["waga"]
        for u, v, dane in G.edges(data=True)
    }

    plt.figure(figsize=(9, 6))
    nx.draw(G, polozenie, with_labels=True, node_color=kolory,
            node_size=1800, arrows=True, arrowsize=22, font_size=12,
            connectionstyle="arc3,rad=0.08")
    nx.draw_networkx_edge_labels(G, polozenie, edge_labels=etykiety_krawedzi,
                                  label_pos=0.35, font_size=10)

    legenda = [
        mpatches.Patch(color="lightgreen", label=f"Wierzchołek {1} (dom)"),
        mpatches.Patch(color="lightcoral", label=f"Wierzchołek {graf.n} (szkoła)"),
        mpatches.Patch(color="lightyellow", label="Pośrednie"),
    ]
    plt.legend(handles=legenda, loc="upper left", fontsize=9)
    plt.title(tytul, fontsize=13)
    plt.axis("off")
    plt.tight_layout()
    plt.show()
    return polozenie


def rysuj_graf_z_wynikiem(graf, wynik_dict, tytul, polozenie=None):
    """Rysuje graf z zaznaczoną najdłuższą ścieżką."""
    G = nx.DiGraph()
    for v in range(1, graf.n + 1):
        G.add_node(v)
    for u, v, w in graf.wszystkie_krawedzie:
        G.add_edge(u, v, waga=w)

    if polozenie is None:
        polozenie = nx.spring_layout(G, seed=42, k=2.0)

    sciezka = wynik_dict.get('sciezka') or []               # krawędzie najdłuższej ścieżki
    krawedzie_sciezki = set()
    if sciezka:
        for i in range(len(sciezka) - 1):
            krawedzie_sciezki.add((sciezka[i], sciezka[i + 1]))

    kolory_krawedzi = []                                    # kolor krawędzi: czerwona jeśli na ścieżce
    for u, v in G.edges():
        if (u, v) in krawedzie_sciezki:
            kolory_krawedzi.append("crimson")
        else:
            kolory_krawedzi.append("#aaaaaa")

    grubosc = []                                            # grubość krawędzi
    for u, v in G.edges():
        grubosc.append(3.5 if (u, v) in krawedzie_sciezki else 1.2)

    kolory_w = []                                           # kolory wierzchołków
    for v in G.nodes():
        if v == 1:
            kolory_w.append("lightgreen")
        elif v == graf.n:
            kolory_w.append("lightcoral")
        elif sciezka and v in sciezka:
            kolory_w.append("gold")                         # złoty = na najdłuższej ścieżce
        else:
            kolory_w.append("lightyellow")

    etykiety = {(u, v): dane["waga"] for u, v, dane in G.edges(data=True)}

    plt.figure(figsize=(9, 6))
    nx.draw(G, polozenie, with_labels=True, node_color=kolory_w,
            node_size=1800, arrows=True, arrowsize=22, font_size=12,
            edge_color=kolory_krawedzi, width=grubosc,
            connectionstyle="arc3,rad=0.08")
    nx.draw_networkx_edge_labels(G, polozenie, edge_labels=etykiety,
                                  label_pos=0.35, font_size=10)

    wynik = wynik_dict['wynik']
    info = f"Wynik: {wynik}"
    if sciezka:
        info += f"\nTrasa: {' → '.join(map(str, sciezka))}"

    legenda = [
        mpatches.Patch(color="lightgreen",  label=f"Wierzchołek {1} (dom)"),
        mpatches.Patch(color="lightcoral",  label=f"Wierzchołek {graf.n} (szkoła)"),
        mpatches.Patch(color="gold",        label="Na najdłuższej ścieżce"),
        mpatches.Patch(color="#aaaaaa",     label="Pozostałe krawędzie"),
        mpatches.Patch(color="crimson",     label="Najdłuższa droga"),
    ]
    plt.legend(handles=legenda, loc="upper left", fontsize=9)
    plt.title(f"{tytul}\n{info}", fontsize=12)
    plt.axis("off")
    plt.tight_layout()
    plt.show()


def rysuj_porzadek_topologiczny(porzadek, tytul):
    """Rysuje wierzchołki ustawione poziomo w porządku topologicznym."""
    if not porzadek:
        return

    fig, ax = plt.subplots(figsize=(max(8, len(porzadek) * 1.5), 2.5))
    ax.set_xlim(-0.5, len(porzadek) - 0.5)
    ax.set_ylim(-0.5, 1.5)

    for i, v in enumerate(porzadek):                       # rysujemy każdy wierzchołek jako kółko
        kolor = "lightgreen" if i == 0 else (
            "lightcoral" if i == len(porzadek) - 1 else "lightyellow"
        )
        circle = plt.Circle((i, 0.5), 0.35, color=kolor,
                              ec="black", zorder=3, lw=1.5)
        ax.add_patch(circle)
        ax.text(i, 0.5, str(v), ha="center", va="center",
                fontsize=13, fontweight="bold", zorder=4)

        if i < len(porzadek) - 1:                          # strzałki między kolejnymi wierzchołkami
            ax.annotate("", xy=(i + 0.65, 0.5), xytext=(i + 0.35, 0.5),
                        arrowprops=dict(arrowstyle="->", color="steelblue", lw=1.5))

    ax.set_title(f"{tytul}\nPorządek topologiczny (algorytm Kahna)", fontsize=12)
    ax.axis("off")
    plt.tight_layout()
    plt.show()


# WYPISYWANIE WYNIKÓW

def wypisz_wynik(nazwa, wynik_dict, obserwacje=None):
    """Wypisuje końcowy wynik i opcjonalne obserwacje."""
    print("\n  ── Wynik ─────────────────────────────")
    w = wynik_dict['wynik']
    if w == 'BRAK':
        print("  BRAK — brak drogi z domu do szkoły.")
    elif w == 'WAGARY':
        print("  WAGARY — istnieje nieskończenie długa droga!")
    else:
        print(f"  Maksymalna długość drogi: {w} m")
        if wynik_dict['sciezka']:
            print(f"  Najdłuższa trasa: {' → '.join(map(str, wynik_dict['sciezka']))}")

    if obserwacje:
        print("\n  ── Obserwacje ────────────────────────")
        for obs in obserwacje:
            print(f"  • {obs}")
    print()


def pokaz_przyklad(nazwa, dane_tekstowe, obserwacje=None, rysuj=True):
    """
    Uniwersalna funkcja prezentująca jeden przykład.
    Wczytuje dane, rozwiązuje, wypisuje przebieg i wynik, rysuje grafy.
    """
    print("\n" + "═" * 60)
    print(f"  {nazwa}")
    print("═" * 60)

    try:
        g = wczytaj_z_tekstu(dane_tekstowe)                 # wczytujemy graf
    except ValueError as e:
        print(f"  Błąd danych wejściowych: {e}")
        return

    print("\n  Graf wejściowy (lista sąsiedztwa):")
    print(g)

    wynik_dict = rozwiaz(g, verbose=True)                   # rozwiązujemy z pełnym opisem
    wypisz_wynik(nazwa, wynik_dict, obserwacje)

    if rysuj:
        polozenie = rysuj_graf_wejsciowy(
            g, f"{nazwa} — graf wejściowy"
        )
        rysuj_graf_z_wynikiem(
            g, wynik_dict, f"{nazwa} — wynik", polozenie
        )
        if wynik_dict['porzadek_topo']:
            rysuj_porzadek_topologiczny(
                wynik_dict['porzadek_topo'],
                f"{nazwa}"
            )


# PRZYKŁADY

def przyklad_1_z_tresci():
    """
    Przykład wprost z treści zadania.
    Oczekiwany wynik: 700 (droga 1 → 3 → 2 → 4).
    """
    dane = """
4 6
1 2 100
1 3 200
2 1 100
2 4 200
3 2 300
4 2 200
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 1: Treść zadania — najdłuższa droga 700 m",
        dane,
        obserwacje=[
            "Oczekiwany wynik: 700.",
            "Droga 1 → 3 → 2 → 4: 200 + 300 + 200 = 700.",
            "Droga 1 → 2 → 4: 100 + 200 = 300 — znacznie krótsza.",
            "Graf jest DAG-iem (krawędź 2 → 1 nie wchodzi do podgrafu bez wierzchołka 1).",
        ],
    )


def przyklad_2_wagary_cykl():
    """
    Przykład z treści: cykl 2-3-4-2, WAGARY.
    """
    dane = """
5 8
1 2 100
1 3 200
2 1 100
2 3 100
2 5 200
3 4 100
4 2 100
5 2 200
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 2: WAGARY — cykl 2 → 3 → 4 → 2",
        dane,
        obserwacje=[
            "Cykl 2 → 3 → 4 → 2 jest osiągalny z wierzchołka 1.",
            "Jarek może krążyć w nieskończoność bez wracania do domu.",
            "Program zwraca WAGARY, mimo że do szkoły (5) da się dojechać.",
        ],
    )


def przyklad_3_wagary_krotki_cykl():
    """
    Przykład z treści: krótki cykl 2-3-4-2, WAGARY.
    """
    dane = """
5 5
1 2 100
2 3 100
3 4 100
4 2 100
1 5 100
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 3: WAGARY — cykl 2 → 3 → 4 → 2 (wariant)",
        dane,
        obserwacje=[
            "Mimo że istnieje prosta droga 1 → 5, to cykl 2 → 3 → 4 → 2 daje WAGARY.",
            "Warunek WAGARY jest sprawdzany niezależnie od tego, czy droga do szkoły istnieje.",
            "Cykl jest osiągalny z 1 (przez krawędź 1 → 2).",
        ],
    )


def przyklad_4_brak():
    """
    Przykład z treści: brak drogi do szkoły, BRAK.
    """
    dane = """
5 7
1 2 100
1 3 200
2 1 100
2 3 100
3 4 100
4 2 100
5 2 200
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 4: BRAK — brak drogi do szkoły (wierzchołek 5)",
        dane,
        obserwacje=[
            "Wierzchołek 5 jest w osobnej składowej — nie można do niego dotrzeć z 1.",
            "BFS z wierzchołka 1 nie dosięga 5 → wynik BRAK.",
            "Istnieje cykl 2 → 3 → 4 → 2, ale skoro nie ma drogi do szkoły, to BRAK.",
            "Kolejność sprawdzania: najpierw osiągalność, potem cykl.",
        ],
    )


def przyklad_5_liniowy():
    """
    Prosta ścieżka liniowa. Jedyna droga, wynik = suma wag.
    """
    dane = """
5 4
1 2 100
2 3 250
3 4 150
4 5 300
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 5: Łańcuch liniowy — jedna droga, wynik = 800 m",
        dane,
        obserwacje=[
            "Jedyna możliwa droga to 1 → 2 → 3 → 4 → 5.",
            "Suma wag: 100 + 250 + 150 + 300 = 800.",
            "DP od razu daje optymalne rozwiązanie w jednym przebiegu.",
        ],
    )


def przyklad_6_wiele_sciezek():
    """
    Wiele alternatywnych ścieżek, algorytm wybiera najdłuższą.
    """
    dane = """
6 9
1 2 10
1 3 50
2 4 20
3 4 30
2 5 100
3 6 200
4 6 80
5 6 40
1 6 5
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 6: Wiele alternatywnych ścieżek",
        dane,
        obserwacje=[
            "Ścieżka 1 → 3 → 6: 50 + 200 = 250.",
            "Ścieżka 1 → 2 → 5 → 6: 10 + 100 + 40 = 150.",
            "Ścieżka 1 → 3 → 4 → 6: 50 + 30 + 80 = 160.",
            "DP wybiera ścieżkę 1 → 3 → 6 = 250 jako najdłuższą.",
        ],
    )


def przyklad_7_izolowany_wierzcholek():
    """
    Szkoła jest izolowana — nie można do niej dotrzeć.
    """
    dane = """
3 1
1 2 500
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 7: Szkoła (wierzchołek 3) jest izolowana — BRAK",
        dane,
        obserwacje=[
            "Istnieje krawędź 1 → 2, ale nie ma żadnej krawędzi do wierzchołka 3.",
            "BFS z 1 odwiedza {1, 2}, nie dochodząc do 3.",
            "Wynik: BRAK.",
        ],
    )


def przyklad_8_rozgaleziony_dag():
    """
    Rozbudowany DAG z wieloma ścieżkami i wąskim gardłem.
    """
    dane = """
7 10
1 2 400
1 3 100
2 4 200
2 5 600
3 5 300
4 6 100
5 6 50
5 7 500
4 7 800
6 7 200
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 8: Rozgałęziony DAG — wąskie gardło",
        dane,
        obserwacje=[
            "Najdłuższa ścieżka: 1 → 2 → 4 → 7: 400 + 200 + 800 = 1400.",
            "DP systematycznie rozważa wszystkie ścieżki w porządku topo.",
            "Wąskie gardło w klasycznym sensie nie spowalnia tu DP — to graf acykliczny.",
        ],
    )


def przyklad_9_jeden_wierzcholek_posredni():
    """
    Małe n = 2, bezpośrednia krawędź z domu do szkoły.
    """
    dane = """
2 1
1 2 999
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 9: n = 2, bezpośrednia krawędź — wynik = 999",
        dane,
        obserwacje=[
            "Najprostszy możliwy przypadek: graf z dwoma wierzchołkami.",
            "Jedna krawędź, jedna droga, wynik = 999.",
        ],
    )


def przyklad_10_wagary_bez_sciezki_do_celu():
    """
    Cykl osiągalny z 1, ale szkoła jest dostępna — mimo to WAGARY.
    Ciekawy przypadek: spełniony jest warunek drogi do szkoły I cyklu.
    """
    dane = """
6 7
1 2 100
2 3 100
3 2 100
2 5 50
5 6 50
1 4 200
4 6 100
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 10: Cykl 2 ↔ 3 i dostępna szkoła — WAGARY",
        dane,
        obserwacje=[
            "Cykl 2 → 3 → 2 jest osiągalny z 1 (przez krawędź 1 → 2).",
            "Szkoła (6) jest dostępna przez 1 → 4 → 6 lub 1 → 2 → 5 → 6.",
            "Mimo że istnieje skończona droga do szkoły, cykl oznacza WAGARY.",
            "Zgodnie z treścią zadania: WAGARY ma pierwszeństwo nad wypisaniem długości.",
        ],
    )

def przyklad_11_8_wierzcholkow():
    dane = """
8 11
1 2 100
1 3 200
2 4 250
2 5 300
3 5 350
3 6 120
5 4 180
5 7 220
6 7 410
4 8 150
7 8 400
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 11: Miasto Jarka — 8 wierzchołków, 11 krawędzi",
        dane,
        obserwacje=[
            "Graf ma 8 wierzchołków.",
            "DP rozważa wszystkie 6 tras i wybiera tę najdłuższą.",
        ],
        rysuj=True,
    )

def przyklad_12_blad_ujemnej_wagi():
    """
    Sprawdźmy reakcję algorytmu na błędne dane - ujemna waga.
    """
    dane = """
2 1
1 2 -100
""".strip()
    pokaz_przyklad(
        "PRZYKŁAD 12: Błędne dane.",
        dane,
        obserwacje=[
        ],
    )

# PORÓWNANIE ZŁOŻONOŚCI I EKSPERYMENT

def eksperyment_wydajnosci():
    """
    Eksperyment ilustrujący liniową złożoność algorytmu.
    Mierzymy czas działania dla rosnących grafów DAG.
    """
    import time
    import random

    print("\n" + "═" * 60)
    print("  EKSPERYMENT: Złożoność algorytmu")
    print("═" * 60)
    print(
        "  Testujemy na losowych DAG-ach o rosnącej liczbie wierzchołków."
        "\n  Krawędź u → v istnieje (p = 0.15) tylko gdy u < v (gwarancja DAG).\n"
    )

    random.seed(2026)                                       # powtarzalność wyników
    wyniki = []

    for n in [100, 500, 1000, 2000, 4000, 6000]:            # rosnące n
        g = Graf(n)
        for u in range(1, n + 1):                           # losujemy krawędzie "do przodu"
            for v in range(u + 1, min(u + 20, n + 1)):      # ograniczamy zasięg dla gęstości
                if random.random() < 0.15:
                    g.dodaj_krawedz(u, v, random.randint(1, 1000))
        if not g.krawedzie[n]:                              # upewniamy się, że n jest osiągalne
            g.dodaj_krawedz(n - 1, n, 1)

        t0 = time.time()
        wynik = rozwiaz(g, verbose=False)                   # cicha wersja
        czas = time.time() - t0

        wyniki.append((n, len(g.wszystkie_krawedzie), wynik['wynik'], czas))
        print(f"  n={n:5d}, m={len(g.wszystkie_krawedzie):6d},"
              f"  wynik={str(wynik['wynik'])[:10]:>10},  czas={czas:.4f}s")

    # Wykres czasu
    ns = [r[0] for r in wyniki]
    czasy = [r[3] for r in wyniki]
    plt.figure(figsize=(8, 4))
    plt.plot(ns, czasy, 'o-', color="steelblue", lw=2, ms=7)
    plt.xlabel("Liczba wierzchołków n")
    plt.ylabel("Czas [s]")
    plt.title("Złożoność algorytmu — czas vs. rozmiar grafu (DAG)\n"
              "O(n + m): czas rośnie liniowo")
    plt.grid(True, alpha=0.4)
    plt.tight_layout()
    plt.show()


# URUCHOMIENIE

if __name__ == "__main__":
    sys.setrecursionlimit(20000)        # zwiększamy limit rekurencji dla dużych grafów

    przyklad_1_z_tresci()
    przyklad_2_wagary_cykl()
    przyklad_3_wagary_krotki_cykl()
    przyklad_4_brak()
    przyklad_5_liniowy()
    przyklad_6_wiele_sciezek()
    przyklad_7_izolowany_wierzcholek()
    przyklad_8_rozgaleziony_dag()
    przyklad_9_jeden_wierzcholek_posredni()
    przyklad_10_wagary_bez_sciezki_do_celu()
    przyklad_11_8_wierzcholkow()
    przyklad_12_blad_ujemnej_wagi()

    eksperyment_wydajnosci()
```
