```python
#Zadanie1
%pip install pulp
import pulp

items = {
    "item1": {"weight": 3.5, "value": 20*3.5},
    "item2": {"weight": 2, "value": 15*2},
    "item3": {"weight": 3, "value": 10*3},
    "item4": {"weight": 5, "value": 5*5},
    "item5": {"weight": 1, "value": 30*1},
    "item6": {"weight": 2.5, "value": 18*2.5}
}
capacity = 10

# Utworzenie problemu optymalizacyjnego
prob = pulp.LpProblem("Continuous Knapsack Problem", pulp.LpMaximize)

# Zmienne decyzyjne
# x[item_name] reprezentuje ułamek (lub ilość) każdego przedmiotu, który jest brany
x = pulp.LpVariable.dicts("x", items.keys(), lowBound=0, upBound=1, cat='Continuous')

# Funkcja celu: maksymalizacja sumy wartości wybranych przedmiotów
prob += pulp.lpSum([items[i]['value'] * x[i] for i in items.keys()]), "Total Value"

# Ograniczenie: całkowita waga nie może przekroczyć pojemności plecaka
prob += pulp.lpSum([items[i]['weight'] * x[i] for i in items.keys()]) <= capacity, "Capacity Constraint"

# Rozwiązanie problemu
prob.solve()

# Wyświetlanie wyników
print("Status:", pulp.LpStatus[prob.status])
print("\nWybrane przedmioty i ich ilości:")
for item_name in items.keys():
    if x[item_name].varValue >= 0:
        print(f"  {item_name}: {x[item_name].varValue:.2f}")

print(f"\nCałkowita waga: {sum([items[i]['weight'] * x[i].varValue for i in items.keys()]):.2f} kg")
print(f"Całkowita wartość: {pulp.value(prob.objective):.2f} zł")
```

```python
#Zadanie2
import pulp

items = {
    "item1": {"weight": 1,   "value": 5000, "max_qty": 1},  
    "item2": {"weight": 0.5, "value": 1500, "max_qty": 3},  
    "item3": {"weight": 0.3, "value": 200,  "max_qty": 5},  
    "item4": {"weight": 4,   "value": 1000, "max_qty": 1},  
    "item5": {"weight": 1,   "value": 500,  "max_qty": 5},  
    "item6": {"weight": 3,   "value": 600,  "max_qty": 2},  
}

capacity = 11.5

# Utworzenie problemu optymalizacyjnego
prob = pulp.LpProblem("Bounded_Knapsack", pulp.LpMaximize)

# Zmienne decyzyjne - ile sztuk każdego przedmiotu bierzemy (liczby całkowite)
x = {
    i: pulp.LpVariable(i, lowBound=0, upBound=items[i]["max_qty"], cat="Integer")
    for i in items
}

# Funkcja celu 
prob += pulp.lpSum(items[i]["value"] * x[i] for i in items)

# Ograniczenie wagowe
prob += pulp.lpSum(items[i]["weight"] * x[i] for i in items) <= capacity


prob.solve()

print("Status:", pulp.LpStatus[prob.status])
print("\nWybrane ilości:")

total_weight = 0
total_value = 0

for i in items:
    qty = int(x[i].varValue)
    print(f"{i}: {qty} szt.")
    total_weight += qty * items[i]["weight"]
    total_value += qty * items[i]["value"]

print(f"\nŁączna waga: {total_weight:.2f} kg")
print(f"Łączna wartość: {total_value:.2f} zł")
```
