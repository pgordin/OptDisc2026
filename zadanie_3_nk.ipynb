```python
import random


# PREORDER I POSTORDER


def preorder(tree, v, parent=None):
    print(v, end=" ")
    for u in tree[v]:
        if u != parent:
            preorder(tree, u, v)


def postorder(tree, v, parent=None):
    for u in tree[v]:
        if u != parent:
            postorder(tree, u, v)
    print(v, end=" ")



# DRZEWO Z KODU PRUFERA


def prufer_to_tree(prufer):
    n = len(prufer) + 2
    degree = [1] * n

    # zwiększamy stopnie według kodu Prüfera
    for v in prufer:
        degree[v] += 1

    tree = {i: [] for i in range(n)}

    for v in prufer:
        for u in range(n):
            if degree[u] == 1:
                tree[u].append(v)
                tree[v].append(u)
                degree[u] -= 1
                degree[v] -= 1
                break

    # ostatnie dwa wierzchołki
    u, v = [i for i in range(n) if degree[i] == 1]
    tree[u].append(v)
    tree[v].append(u)

    return tree



# LOSOWY GRAF DWUDZIELNY


def random_bipartite_graph(n, p):
    graph = {i: [] for i in range(2 * n)}

    for u in range(n):           # zbiór A
        for v in range(n, 2 * n):  # zbiór B
            if random.random() < p:
                graph[u].append(v)
                graph[v].append(u)

    return graph



# TESTY / PRZYKŁADY


if __name__ == "__main__":
    print("=== DRZEWO Z KODU PRÜFERA ===")
    prufer = [1, 1, 2]  # przykładowy kod
    tree = prufer_to_tree(prufer)

    print("Drzewo (lista sąsiedztwa):")
    for v in tree:
        print(v, ":", tree[v])

    print("\nPreorder:")
    preorder(tree, 0)

    print("\nPostorder:")
    postorder(tree, 0)

    print("\n\n=== LOSOWY GRAF DWUDZIELNY ===")
    g = random_bipartite_graph(4, 0.5)

    for v in g:
        print(v, ":", g[v])
```

```python
import random

def random_bipartite_graph(n, p):
    """
    Generuje losowy graf dwudzielny jako podgraf K_{n,n}.

    Parametry:
    n - liczba wierzchołków w każdej części (łącznie 2n)
    p - prawdopodobieństwo dodania każdej krawędzi

    Zwraca:
    graf w postaci listy sąsiedztwa (słownik)
    """

    # inicjalizacja pustego grafu
    graph = {i: [] for i in range(2 * n)}

    # zbiór A: 0 ... n-1
    # zbiór B: n ... 2n-1
    for u in range(n):
        for v in range(n, 2 * n):
            if random.random() < p:
                graph[u].append(v)
                graph[v].append(u)

    return graph


# PRZYKŁADOWE URUCHOMIENIE


if __name__ == "__main__":
    n = 4
    p = 0.5

    g = random_bipartite_graph(n, p)

    print("Graf (lista sąsiedztwa):")
    for v in g:
        print(v, ":", g[v])
```
