```python
from collections import deque

class Graph:
    def __init__(self):
        # Słownik przechowujący graf w postaci list sąsiedztwa:
        # klucz = zadanie, wartość = lista zadań zależnych od niego
        self.graph = {}

    def add_vertex(self, vertex):
        # Dodawanie wierzchołka (zadania do grafu), jeśli jeszcze nie istnieje
        if vertex not in self.graph:
            self.graph[vertex] = []

    def add_edge(self, u, v):
        # Krawędź skierowana z u do v
        # u -> v oznacza, że zadanie u musi zostać wykonane przed rozpoczęciem zadania v
        self.add_vertex(u)
        self.add_vertex(v)
        self.graph[u].append(v) # Dodanie zadania zależnego do listy sąsiedztwa

    def topological_sort_kahn(self):
        """Implementacja Algorytmu Kahna"""
        #Algorytm wybiera zadania, które nie mają żadnych niewykonanych poprzedników
        #(stopień wejściowy = 0).
        
        # Obliczanie stopni wejściowych (Dla każdego wierzchołka zapisujemy liczbę 
        # krawędzi wchodzących)
        in_degree = {vertex: 0 for vertex in self.graph}

        # Obliczanie stopni wejściowych:
        for vertex in self.graph:
            for neighbor in self.graph[vertex]:
                in_degree[neighbor] += 1 # Zwiększamy stopień wejściowy każdego sąsiada

        # Kolejka z zadaniami, które nie mają żadnych poprzedników (stopień = 0), zostają wykonane od razu
        # Do kolejki trafiają zadania, które można wykonać od razu
        queue = deque([v for v in in_degree if in_degree[v] == 0])
        # Lista przechowująca końcową kolejność zadań
        result = []

        while queue:
            # Wykonujemy zadania, które nie mają niewykonanych poprzedników
            current = queue.popleft()
            # Dodanie zadania do końcowej kolejności wykonywania
            result.append(current)

            for neighbor in self.graph[current]:
                # Usuwamy wpływ właśnie wykonanego zadania
                in_degree[neighbor] -= 1
                # Zadanie nie ma już żadnych oczekujących zależności:
                if in_degree[neighbor] == 0:
                    queue.append(neighbor)
                    
        # Jeśli nie udało się odwiedzić wszystkich zadań, oznacza to istnienie cyklu zależności
        if len(result) != len(self.graph):
            return ["Błąd: Graf zawiera cykl! Nie można ułożyć harmonogramu."]

        return result

    def topological_sort_dfs(self):
        """Implementacja Algorytmu DFS"""
        # Wierzchołek trafia na stos dopiero wtedy,gdy wszystkie zadania od niego zależne
        #zostały już odwiedzone.
        
        visited = set() # Zbiór odwiedzonych wierzchołków
        stack = [] # Stos przechowujący wynik

        def dfs(vertex): # Rekurencyjne przejście grafu w głąb
            visited.add(vertex)# Oznaczenie zadania jako odwiedzonego
            
           # Przechodzimy do wszystkich sąsiadów bieżącego wierzchołka
            for neighbor in self.graph.get(vertex, []):
                if neighbor not in visited:
                    # Odwiedzamy sąsiada tylko raz
                    dfs(neighbor)
            stack.append(vertex) # Dodajemy zadanie na stos po odwiedzeniu wszystkich jego następców

        for vertex in self.graph:
            if vertex not in visited:
                dfs(vertex)

        return stack[::-1] # Odwrócenie stosu daje poprawną kolejność wykonywania zadań
```

```python
print("=== PRZYKŁAD 1: PRZYGOTOWANIE OBIADU ===")
g_obiad = Graph() # Tworzymy graf zależności pomiędzy zadaniami
g_obiad.add_edge("Zakupy", "Przygotowanie_skladnikow")
g_obiad.add_edge("Przygotowanie_skladnikow", "Gotowanie")
g_obiad.add_edge("Gotowanie", "Podanie_obiadu")
g_obiad.add_edge("Nakrycie_stolu", "Podanie_obiadu")

print("Algorytm Kahna:")
for task in g_obiad.topological_sort_kahn():
    print(f" - {task}")

print("\nAlgorytm DFS:")
for task in g_obiad.topological_sort_dfs():
    print(f" - {task}")
```
