```python
#Największe skojarzenie w grafach dwudzielnych
```

```python
class BipartiteMatching:

    def __init__(self):
                                                     #dziewczyna -> lista wybranych chłopcow
        self.graf = {}

    def dodaj_wybor(self, dziewczyna, chlopiec):     #dodaje informacje o tym, ktorego chlopca wybrala dana dziewczyna

        if dziewczyna not in self.graf:
            self.graf[dziewczyna] = []

        self.graf[dziewczyna].append(chlopiec)

    def dfs(self, dziewczyna, odwiedzone):           #szuka mozliwosci przypisania chlopca do dziewczyny (czy jest wolny)

        for chlopiec in self.graf.get(dziewczyna, []):

            if chlopiec in odwiedzone:
                continue

            odwiedzone.add(chlopiec)

            if chlopiec not in self.para_chlopiec:

                self.para_chlopiec[chlopiec] = dziewczyna
                return True

            poprzednia_dziewczyna = self.para_chlopiec[chlopiec]

            if self.dfs(poprzednia_dziewczyna, odwiedzone):

                self.para_chlopiec[chlopiec] = dziewczyna
                return True

        return False

    def najwieksze_skojarzenie(self):     #uruchamia algorytm dla kazdej dziewczyny i stopniowo zwieksza liczbe znalezionych par

        self.para_chlopiec = {}

        licznik = 0

        for dziewczyna in self.graf:

            odwiedzone = set()

            if self.dfs(dziewczyna, odwiedzone):
                licznik += 1

        return licznik, self.para_chlopiec
```

```python
#przyklad 1 - pelne skojarzenie

bm = BipartiteMatching()

bm.dodaj_wybor("D1", "C1")
bm.dodaj_wybor("D2", "C2")
bm.dodaj_wybor("D3", "C3")

liczba_par, wynik = bm.najwieksze_skojarzenie()

print("Liczba par:", liczba_par)

for chlopiec, dziewczyna in wynik.items():
    print(dziewczyna, "-", chlopiec)
```
